```python
color_map = {
    "FedBis":   "#1f77b4",  # blue      -> Single
    "Plain":    "#d62728",  # red       -> Hard
    "Oracle":   "#ff7f0e",  # orange    -> Hard-Oracle
    "Soft":     "#2ca02c",  # green     -> Soft
    "Base":           "#9467bd",  # purple    -> Base
    "Centralize":     "#e377c2",  # pink      -> pink
}


# --- legend에 표시될 이름 ---
legend_map = {
    "FedBis": "Single",
    "Plain": "Hard",
    "Oracle": "Hard-Oracle",
    "Soft": "Soft",
    "Base": "Base",
    "Centralize": "Centralize",   
}




select_steps = [0, 80, 160, 240]  



def plot_panel(ax, filtered, title):
    series_keys = [
        "FedBis",
        "Plain",
        "Oracle",
        "Soft",
        "Base",
        "Centralize",
    ]

    for key in series_keys:
        ys = [filtered[key][s] for s in select_steps]

        ax.plot(
            select_steps,
            ys,
            marker="o",
            linewidth=1.0,
            markersize=4,
            label=legend_map[key],      # legend에 들어갈 이름
            color=color_map[key],       # 위에서 정의한 색
        )

    ax.set_title(title)
    ax.set_xlabel("Fine-tuning steps")
    ax.set_xticks(select_steps)
    ax.grid(True, alpha=0.3)




fig, axes = plt.subplots(1, 2, figsize=(10, 4))

plot_panel(axes[0], filtered_real_total, "Real dataset")
plot_panel(axes[1], filtered_syn_total,  "Synthetic dataset")
axes[0].set_ylabel("Mean Test Accuracy")

handles, labels = axes[0].get_legend_handles_labels()


fig.legend(
    handles, labels,
    loc="upper center",
    ncol=6,
    bbox_to_anchor=(0.5, 1.02),  # 🔹 1.08 → 1.02 로 줄여서 더 타이트하게
)

plt.tight_layout()
plt.subplots_adjust(top=0.82)     # 🔹 0.78 정도였으면 0.82~0.85로 올려보기
plt.savefig("./final_combined_real_synthetic_fed_alg_comparison_total.png",
            dpi=300, bbox_inches="tight")
plt.show()


```
