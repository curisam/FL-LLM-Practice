```python
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import zoomed_inset_axes, mark_inset, InsetPosition

steps_full   = list(range(0, 301, 10))
select_steps = [0, 100, 200, 300]

color_map = {
    1: "#1f77b4",
    3: "#d62728",
    4: "#ff7f0e",
    5: "#2ca02c",
}

def plot_panel(ax, filtered, title):
    series_info = [
        ("Single (K=1)", 1, "FedBis (C=1)"),
        ("Hard-Balance (K=3)",   3, "Balancing (C=3)"),
        ("Hard-Balance (K=4)",   4, "Balancing (C=4)"),
        ("Hard-Balance (K=5)",   5, "Balancing (C=5)"),
    ]
    for name, k_val, key in series_info:
        ys = [filtered[key][s] for s in select_steps]
        ax.plot(
            select_steps,
            ys,
            marker="o",
            linewidth=1.0,
            markersize=4,
            label=name,
            color=color_map[k_val],
        )

    ax.set_title(title)
    # ax.set_xlabel("Fine-tuning steps")
    ax.set_xticks(select_steps)   # ⬅ 여기 한 줄만 추가
    ax.grid(True, alpha=0.35)

plt.rcParams.update({
    "font.size": 12,
    "axes.titlesize": 14,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "legend.fontsize": 11,
})

# ▶ 세로 사이즈 조금 키우고, 패널 사이 간격(hspace) 줄이기
fig, axes = plt.subplots(
    2, 1,
    figsize=(6, 8.5),
    gridspec_kw={"hspace": 0.25}
)

axes[1].set_xlabel("Fine-tuning steps")
axes[1].set_xticks(select_steps)   # ⬅ 여기 한 줄만 추가
axes[0].set_xticks(select_steps)   # ⬅ 여기 한 줄만 추가

plot_panel(axes[0], filtered_real, "Real dataset")
plot_panel(axes[1], filtered_syn,  "Synthetic dataset")
# axes[0].set_ylabel("Mean Test Accuracy")
# axes[1].set_ylabel("Mean Test Accuracy")

axes[1].set_xlabel("Fine-tuning steps")

# ===== Synthetic 축 안에 zoom 박스 =====
ax_syn = axes[1]
ys_k3 = [filtered_syn["Balancing (C=3)"][s] for s in select_steps]
ys_k5 = [filtered_syn["Balancing (C=5)"][s] for s in select_steps]
ys_k4 = [filtered_syn["Balancing (C=4)"][s] for s in select_steps]
ymin = min(min(ys_k3), min(ys_k5), min(ys_k4))
ymax = max(max(ys_k3), max(ys_k5), max(ys_k4))
margin = (ymax - ymin) * 0.2 if ymax > ymin else 0.01

ax_ins = zoomed_inset_axes(ax_syn, zoom=2.5, loc="upper right")
ax_ins.set_box_aspect(1.0)
ip = InsetPosition(ax_syn, [0.55, 0.2, 0.4, 0.4])
ax_ins.set_axes_locator(ip)
ax_ins.plot(select_steps, ys_k3, marker="o", linewidth=1.0, markersize=4, color=color_map[3])
ax_ins.plot(select_steps, ys_k4, marker="o", linewidth=1.0, markersize=4, color=color_map[4])
ax_ins.plot(select_steps, ys_k5, marker="o", linewidth=1.0, markersize=4, color=color_map[5])


x_min, x_max = min(select_steps), max(select_steps)
x_range = x_max - x_min
x_margin = x_range * 0.05

ax_ins.set_xlim(x_min - x_margin, x_max + x_margin)
ax_ins.set_ylim(ymin - margin, ymax + margin)

# ▶ inset 축에서는 눈금 안 보이게
ax_ins.set_xticks([])
# ax_ins.set_yticks([])

mark_inset(ax_syn, ax_ins, loc1=2, loc2=4, fc="none", ec="0.4", lw=0.8)

# ===== 공통 legend : 플롯에 더 가깝게 =====
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels,
           loc="lower center",
           ncol=2,
           bbox_to_anchor=(0.5, 0.08))   # y=0.03 → 0.08 : 플롯 쪽으로 당김

# legend 자리 확보를 위해 bottom만 살짝 여유
plt.tight_layout()
plt.subplots_adjust(bottom=0.25)

plt.savefig("./final_combined_real_synthetic_fed_alg_comparison_final_style_balance.pdf",
            dpi=300, bbox_inches="tight")
plt.show()
```
