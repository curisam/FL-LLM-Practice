```python
# --- renderer error 우회 (노트북에서 plt 렌더러가 깨진 경우에도 보이게) ---
import matplotlib
matplotlib.use("Agg")

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.ticker import MaxNLocator, MultipleLocator
from IPython.display import Image, display

# -----------------------
# Colors (유지)
# -----------------------
COLOR_SINGLE = "#1f77b4"
COLOR_CENT   = "#e377c2"

# x: 등간격 배치 + 라벨
x_labels = [1, 5, 15, 30]
x_pos = np.arange(len(x_labels))  # 0,1,2,3

def extract_y_percent(d, finetune_step):
    return [100.0 * d[k][finetune_step] for k in x_labels]

def auto_ylim(y_vals, pad_ratio=0.12, min_pad=1.5):
    y_min = float(min(y_vals))
    y_max = float(max(y_vals))
    rng = max(y_max - y_min, 1e-9)
    pad = max(rng * pad_ratio, min_pad)
    return (y_min - pad, y_max + pad)

def plot_2x2_lr_ft(out_png="lr_ft_2x2.png"):
    # -----------------------
    # Data (percent)
    # -----------------------
    # High lr (origin) -> filled circle
    high0_single  = extract_y_percent(single_origin,      0)
    high0_cent    = extract_y_percent(centralize_origin,  0)
    high80_single = extract_y_percent(single_origin,     80)
    high80_cent   = extract_y_percent(centralize_origin, 80)

    # Smaller lr (small) -> hollow circle (edge only)
    small0_single  = extract_y_percent(single_small,      0)
    small0_cent    = extract_y_percent(centralize_small,  0)
    small80_single = extract_y_percent(single_small,     80)
    small80_cent   = extract_y_percent(centralize_small, 80)

    # -----------------------
    # Figure layout: 2x2
    # rows: High lr / Smaller lr
    # cols: ft=0 / ft=80
    # -----------------------
    fig, axes = plt.subplots(2, 2, figsize=(11.2, 7.2), sharex=True)

    # 같은 x에서 Single/Centralize 겹침 방지
    dx = 0.10
    xs_single = x_pos - dx
    xs_cent   = x_pos + dx

    s = 220
    lw = 2.2

    # -----------------------
    # (row 0) High lr : filled circle
    # -----------------------
    ax = axes[0, 0]
    ax.scatter(xs_single, high0_single, s=s, marker="o",
               facecolors=COLOR_SINGLE, edgecolors=COLOR_SINGLE, linewidths=lw, zorder=3)
    ax.scatter(xs_cent,   high0_cent,   s=s, marker="o",
               facecolors=COLOR_CENT, edgecolors=COLOR_CENT, linewidths=lw, zorder=3)
    ax.set_title("Before Personalization")
    ax.set_ylabel("Accuracy (%)")
    ax.set_ylim(*auto_ylim(high0_single + high0_cent, pad_ratio=0.15, min_pad=0.8))

    ax = axes[0, 1]
    ax.scatter(xs_single, high80_single, s=s, marker="o",
               facecolors=COLOR_SINGLE, edgecolors=COLOR_SINGLE, linewidths=lw, zorder=3)
    ax.scatter(xs_cent,   high80_cent,   s=s, marker="o",
               facecolors=COLOR_CENT, edgecolors=COLOR_CENT, linewidths=lw, zorder=3)
    ax.set_title("After Personalization")
    # (오른쪽은 y-range 고정할 거라 여기서 ylim은 아래에서 덮어씀)

    # -----------------------
    # (row 1) Smaller lr : hollow circle (edge only, solid)
    # -----------------------
    ax = axes[1, 0]
    ax.scatter(xs_single, small0_single, s=s, marker="o",
               facecolors="none", edgecolors=COLOR_SINGLE, linewidths=lw, zorder=3)
    ax.scatter(xs_cent,   small0_cent,   s=s, marker="o",
               facecolors="none", edgecolors=COLOR_CENT, linewidths=lw, zorder=3)
    ax.set_ylabel("Accuracy (%)")
    ax.set_ylim(*auto_ylim(small0_single + small0_cent, pad_ratio=0.15, min_pad=0.8))

    ax = axes[1, 1]
    ax.scatter(xs_single, small80_single, s=s, marker="o",
               facecolors="none", edgecolors=COLOR_SINGLE, linewidths=lw, zorder=3)
    ax.scatter(xs_cent,   small80_cent,   s=s, marker="o",
               facecolors="none", edgecolors=COLOR_CENT, linewidths=lw, zorder=3)
    # (오른쪽은 y-range 고정할 거라 여기서 ylim은 아래에서 덮어씀)

    # -----------------------
    # Common x formatting (등간격 + 라벨)
    # - 두 행 모두 숫자(틱 라벨)는 보이게
    # - x축 제목("Local iteration steps")은 아래 행만
    # -----------------------
    for r in range(2):
        for c in range(2):
            ax = axes[r, c]
            ax.set_xticks(x_pos)
            ax.set_xticklabels([str(v) for v in x_labels])
            ax.set_xlim(-0.5, len(x_labels) - 0.5)
            ax.tick_params(axis="x", which="both", labelbottom=True)

            if r == 1:
                ax.set_xlabel("Local iteration steps")
            else:
                ax.set_xlabel("")

    # =========================================================
    # ✅ 요청 반영: y-tick/grid 수정
    # 1) 왼쪽 컬럼: 48.5 같은 0.5 tick 제거(정수 tick만)
    # 2) 오른쪽 컬럼: 50~70으로 고정 + 10 단위 tick/grid
    # =========================================================
    # (1) 왼쪽 컬럼(Before): 정수 tick만
    for ax in [axes[0, 0], axes[1, 0]]:
        ax.yaxis.set_major_locator(MaxNLocator(integer=True))

    # (2) 오른쪽 컬럼(After): 50~100 + 10 단위 tick/grid + 마커 안 잘리게 여유
    for ax in [axes[0, 1], axes[1, 1]]:
        ax.set_ylim(49, 101)  # ✅ 약간 여유를 줘서 마커가 위아래로 안 잘리게
        ax.set_yticks([50, 60, 70, 80, 90, 100])  # ✅ 원하는 tick 고정
        ax.yaxis.set_major_locator(MultipleLocator(10))
        ax.grid(False, axis="y", which="major")

    # -----------------------
    # Unified legend: 4 items (row-wise order)
    # 1행: Single original, Centralize original
    # 2행: Single smaller,  Centralize smaller
    # -----------------------
    legend_handles = [
        Line2D([0], [0], marker="o", linestyle="None", markersize=12,
               markerfacecolor=COLOR_SINGLE, markeredgecolor=COLOR_SINGLE,
               markeredgewidth=lw, label="Single (original lr)"),
        Line2D([0], [0], marker="o", linestyle="None", markersize=12,
               markerfacecolor="none", markeredgecolor=COLOR_SINGLE,
               markeredgewidth=lw, label="Single (smaller lr)"),
        Line2D([0], [0], marker="o", linestyle="None", markersize=12,
               markerfacecolor=COLOR_CENT, markeredgecolor=COLOR_CENT,
               markeredgewidth=lw, label="Centralize (original lr)"),
        Line2D([0], [0], marker="o", linestyle="None", markersize=12,
               markerfacecolor="none", markeredgecolor=COLOR_CENT,
               markeredgewidth=lw, label="Centralize (smaller lr)"),
    ]

    fig.legend(handles=legend_handles, loc="lower center", ncol=2,
               frameon=True, bbox_to_anchor=(0.5, -0.01))

    fig.tight_layout(rect=[0, 0.06, 1, 1])
    fig.savefig(out_png, dpi=220, bbox_inches="tight")
    plt.close(fig)
    display(Image(filename=out_png))


# 실행
plot_2x2_lr_ft("final_sensitivity_single_centralize.png")
```
